```python
def recommend(movie_title, n=10):
    # پیدا کردن فیلم
    matches = movies_final[
        movies_final['title'].str.lower() == movie_title.lower()
    ]

    if matches.empty:
        return f"Movie '{movie_title}' not found."

    movie_index = matches.index[0]

    # گرفتن similarity مربوط به فیلم
    similarity_scores = list(
        enumerate(similarity[movie_index])
    )

    # مرتب‌سازی بر اساس بیشترین شباهت
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # حذف خود فیلم
    similarity_scores = similarity_scores[1:n+1]

    # گرفتن index فیلم‌ها
    movie_indices = [x[0] for x in similarity_scores]

    # ساخت نتیجه
    recommendations = movies_final.iloc[movie_indices][
        ['title']
    ].copy()

    recommendations['similarity'] = [
        round(x[1], 3)
        for x in similarity_scores
    ]

    return recommendations.reset_index(drop=True)
```

```python
def search_movies(query):
    results = movies_final[
        movies_final['title'].str.contains(
            query,
            case=False,
            na=False
        )
    ]

    return results[['title']].reset_index(drop=True)
```

```python
from difflib import get_close_matches
```

```python
def find_similar_movies(query, limit=10, cutoff=0.5):
    titles = movies_final['title'].tolist()

    matches = get_close_matches(
        query,
        titles,
        n=limit,
        cutoff=cutoff
    )

    return matches
```

```python
def smart_recommend(query, n=10):
    # پیدا کردن فیلم‌هایی که دقیقاً با عبارت واردشده مطابقت دارند
    exact_matches = movies_final[
        movies_final['title'].str.lower() == query.lower()
    ]

    # اگر فیلم دقیقاً پیدا شد
    if not exact_matches.empty:
        movie_title = exact_matches.iloc[0]['title']

        print(f"Movie found: {movie_title}")

        return recommend(movie_title, n)

    # اگر فیلم دقیق پیدا نشد، جستجوی تقریبی انجام می‌دهیم
    similar_titles = find_similar_movies(query, limit=5)

    if not similar_titles:
        return "No similar movies found."

    print("Movie not found exactly")
    print("Did you mean one of these?")

    for i, title in enumerate(similar_titles, start=1):
        print(f"{i}. {title}")

    return None
```

```python
def smart_recommend(query, n=10):
    # جستجوی دقیق
    exact_matches = movies_final[
        movies_final['title'].str.lower() == query.lower()
    ]

    if not exact_matches.empty:
        movie_title = exact_matches.iloc[0]['title']

        print(f"فیلم پیدا شد: {movie_title}")

        return recommend(movie_title, n)

    # جستجوی تقریبی
    similar_titles = find_similar_movies(query, limit=5)

    if not similar_titles:
        return "No similar movies found."

    # بهترین تطابق
    best_match = similar_titles[0]

    print(f"'{query}' پیدا نشد.")
    print(f"مشابه ترین کلمه: {best_match}")

    return recommend(best_match, n)
```
